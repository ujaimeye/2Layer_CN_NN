```python
%run func_defs.ipynb

nz = 400 ; nt = 300
Lz=40 ; tf = 20; 
out_file = f'./data/setB2_nz{nz}_nt{nt}_tf{tf}.pt'
device=torch.device('cuda')
```

```python

#Create the data sets
dz = Lz/nz; dt = tf/nt
z1 = 0.5*dz + torch.arange(0, nz, device=device)*dz
t1 = 0.5*dt + torch.arange(0, nt, device=device)*dt

zt_grid = torch.cartesian_prod(z1, t1)
z = zt_grid[:, 0]; t = zt_grid[:, 1] # (z, t): Long list of grid points

stau = torch.column_stack(z_t_to_s_tau(z, t))
W = W_stau(stau).reshape(-1,1)

data_to_save = {
        't': t,
        'z': z,
        'stau': stau,
        'W': W
    }
torch.save(data_to_save, out_file)
print(f'Data saved in {out_file}')
```
