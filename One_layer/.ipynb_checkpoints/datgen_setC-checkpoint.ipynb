```python
%reset -f
%run func_defs.ipynb

nz = 400 ; nt = 300
Lz=40 ; tf = 20; 
out_file = f'./data/setC_nz{nz}_nt{nt}_tf{tf}.pt'
device=torch.device('cuda')
```

```python

# Create the data sets
s_0 = 1e-12;
ds = 1/nz
tau_0 = 1/(1+tf)
dtau = 1/nt

# grid points (intersections of grid lines)
s1_draft = torch.arange(0, nz+1, device=device)*ds
s1 = s_0 + (1-s_0)*s1_draft
tau1_draft = torch.arange(0, nt+1, device=device)*dtau
tau1 = tau_0 + (1-tau_0)*tau1_draft

stau = torch.cartesian_prod(s1, tau1)
W = W_stau(stau).reshape(-1,1)

s = stau[:, 0]; tau = stau[:, 1] # (s, tau): Long list of grid points
z, t = s_tau_to_z_t(s, tau)

data_to_save = {
        'stau': stau,
        'W': W
    }
torch.save(data_to_save, out_file)
print(f'Data saved in {out_file}')

z2d = z.reshape(-1,len(tau1))
t1 = t.reshape(-1,len(tau1))[0,:]
print(f'at t = {t1[0]:.4e}, (z_min, z_max): ({torch.min(z2d[:,0]):.4e}, {torch.max(z2d[:,0]):.4e})')
print(f'at t = {t1[-1]:.4e}, (z_min, z_max): ({torch.min(z2d[:,-1]):.4e}, {torch.max(z2d[:,-1]):.4e})')
```
